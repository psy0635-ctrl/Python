```python
import numpy as np
import pandas as pd

df = pd.DataFrame({'name': ['Jessica', 'Liam', 'Sophia', 'Ryan', 'Alex'],
                   'gender': ['F', np.nan, np.nan, 'M', 'F'],
                   'score': [89, 75, 95, np.nan, 93]})
display(df)
```

```python
clean_df = df.dropna(subset=['name','score'])
display(clean_df)
```

```python
import pandas as pd

df = pd.DataFrame({'name': ['Jessica', 'Liam', 'Sophia', 'Ryan', 'Alex'],
                   'gender': ['F', np.nan, np.nan, 'M', 'F'],
                   'score': [89, 75, 95, np.nan, 93]})

df1 = df.fillna(0)
display(df1)
```

```python
df1 = df.fillna({'gender':'etc', 'score':0})
display(df1)
```

```python
df.fillna({'gender':'etc', 'score':df['score'].mean()}, inplace=True)
display(df)
```

```python
import pandas as pd

df1 = pd.DataFrame({'name': ['Jessica', 'Liam', 'Sophia', 'Ryan','Alex'],
                    'gender': ['F', 'M', 'F', 'M', 'F'],
                    'age': [21, 23, 20, 22, 20]})

df2 = pd.DataFrame({'name': ['Sophia', 'Liam', 'Ryan', 'Alex', 'Jessica'],
                    'department': ['computer science', 'mathematics', 'law',
                    'computer science', 'law']})
```

```python
df3 = pd.merge(df1, df2)
display(df3)
```

```python
df3 = pd.merge(df1, df2, on='name')
display(df3)
```

```python
df4 = pd.DataFrame({'department': ['computer science', 'mathematics', 'law'],
                    'head of department': ['Steve', 'Emma', 'Carly']})
```

```python
df5 = pd.merge(df3, df4, on='department')
display(df5)
```

```python
import pandas as pd

df = pd.DataFrame({'온도': ['섭씨10도', '20', '15도'],
                   '높이': ['1,250미터', '740', '1,130미터']})
df['온도'] = df['온도'].str.strip() \
                       .str.rstrip('도') \
                       .str.lstrip('섭씨') \
                       .astype(int)
df['높이'] = df['높이'].str.strip() \
                       .str.rstrip('미터') \
                       .str.replace(',','') \
                       .astype(int)
display(df)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('precipitation.csv')
display(df)
```

```python
plt.figure(figsize=(6,4))
plt.bar(df['year'], df['spring'], label='spring')
plt.bar(df['year'], df['summer'], bottom=df['spring'], label='summer')
plt.bar(df['year'], df['autumn'], bottom=(df['spring']+df['summer']), label='autumn')
plt.bar(df['year'], df['winter'], bottom=(df['spring']+df['summer']+df['autumn']), label='winter')
plt.ylabel('precipitation')
plt.xlabel('year')
plt.legend()
plt.show()
```

```python
df['sum'] = df['spring']+df['summer']+df['autumn']+df['winter']
display(df)
```
