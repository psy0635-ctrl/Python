```python
from deep_translator import MyMemoryTranslator

user_input = input("번역할 한글 문장을 입력하세요: ")

translated_en = MyMemoryTranslator(source='ko-KR', target='en-US').translate(user_input)
translated_ja = MyMemoryTranslator(source='ko-KR', target='ja-JP').translate(user_input)

print("\n" + "=" * 40)
print(f"입력(한글) : {user_input}")
print(f"영어(EN)   : {translated_en}")
print(f"일본어(JA) : {translated_ja}")
print("=" * 40)
```

2. 번역기

```python
from deep_translator import MyMemoryTranslator
```

```python
while True:
  lang = input('도착어 선택(영어, 일본어, 독일어, 종료): ')
  if lang=='종료':
    print('종료합니다.')
    break
  elif lang=='영어':
    target_lang = 'en-US'
  elif lang=='일본어':
    target_lang = 'ja-JP'
  elif lang=='독일어':
    target_lang = 'de-DE'
  else:
    print('잘못 입력했습니다.')
    continue
  src = input('번역할 내용을 입력하세요: ')
  if src.strip():
    dest = MyMemoryTranslator(source='ko-KR', target=target_lang).translate(src)
    print(f"-> [{lang} 번역 결과]: {dest}\n")
  else:
    print("입력된 내용이 없습니다.\n")
```
