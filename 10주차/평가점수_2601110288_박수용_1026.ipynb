5명의 위원 평가점수 평균 구하기(최고점수 및 최저점수 제외 기능 추가)

```python

# 5명의 심사위원 평가점수 평균 구하기
judge_num = 5
total_score = 0
max_score = 0
min_score = 0

# 심사위원 점수 입력
for i in range(judge_num):
    score = int(input(f"심사위원{i+1} 평가점수: "))
    total_score += score

    # 최고점수와 최저점수 구하기
    if i == 0:
        max_score = score
        min_score = score
    else:
        if score > max_score:
            max_score = score

        if score < min_score:
            min_score = score

# 전체 평균 구하기
total_avg = total_score / judge_num

# 최고점수와 최저점수 제외한 평균
total_score = total_score - max_score - min_score
avg = total_score / (judge_num - 2)

# 결과 출력
print("최고점수:", max_score)
print("최저점수:", min_score)
print("전체 평가점수 평균:", round(total_avg, 2))
print("최고/최저 제외 평균:", round(avg, 2))
```
