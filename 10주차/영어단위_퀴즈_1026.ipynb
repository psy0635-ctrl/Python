1. 동물 이름 퀴즈

```python
words = {'사자':'lion','호랑이':'tiger','토끼':'rabbit'}
```

```python
for key, value in words.items():
  ans = input(key+' 영어 이름은? ')
  if ans == value:
    print('맞았습니다.')
  else:
    print('틀렸습니다.')
```

2. 영어 단어 퀴즈

```python
from deep_translator import MyMemoryTranslator
translator = MyMemoryTranslator(source='ko-KR', target='en-US')
```

```python
words = {}
```

```python
def registration():
  s = input('등록할 단어: ').strip()
  if not s:
    print('단어를 입력해주세요.')
    return

  if s in words:
    print('이미 등록된 단어입니다.')
  else:
    # MyMemory 번역 실행 (결과 문자열을 직접 반환하므로 .text 없이 바로 처리)
    translated_text = translator.translate(s).strip().lower()
    words[s] = translated_text
    print(f"'{s}' -> '{translated_text}' 등록 완료!")
```

```python
def quiz():
  if not words:
    print('등록된 단어가 없습니다. 먼저 단어를 등록해주세요.')
    return

  for key, value in words.items():
    ans = input(f"'{key}'의 영어 이름은? ").strip().lower()
    if ans == value:
      print('정답입니다! 🎉')
    else:
      print(f'틀렸습니다. (정답: {value})')
```

```python
while True:
  try:
        sel = int(input('\n1.등록  2.퀴즈  3.단어목록확인  4.종료 : '))
  except ValueError:
        print('숫자(1~4)만 입력해주세요.')
        continue

  if sel == 4:
        print('프로그램을 종료합니다.')
        break
  elif sel == 1:
        registration()
  elif sel == 2:
        quiz()
  elif sel == 3:
        if words:
            print('--- 현재 등록된 단어장 ---')
            for k, v in words.items():
                print(f'{k} : {v}')
        else:
            print('단어장이 비어 있습니다.')
  else:
        print('1번부터 4번까지의 메뉴 중에서 선택해주세요.')
```
